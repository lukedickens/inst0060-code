### 1.6 b)
The code below creates a larger set of equally spaced input values (`xs`) using the `numpy.linspace` function. These inputs are then passed as input to the `expand_to_monomials` function and the columns of the resulting output matrix are plotted. Can you explain the shapes of the plots?

```python
num_points = 51
xs = np.linspace(-1, 1, num_points)
degree = 3
expanded_xs = expand_to_monomials(xs, degree)
# iterate over each column of expanded_dummy_inputs
# each column is the values xs to a given power
for k in range(expanded_xs.shape[1]):
    plt.plot(
      xs, expanded_xs[:,k], '-', label=f"$x^{k}$")
plt.legend()
plt.xlabel("x")
plt.ylabel("$y$")
```

**Solution notes** The $j$th column of the output matrix, `expanded_xs`, contains the values `xs` each raised to the $j$th power, so if the input `xs` is $(x_1, x_2, x_3, \ldots, x_N)$ then the $j$th column of `expanded_xs` contains $(x_1^j, x_2^j, x_3^j, \ldots, x_N^j)$. As the values in `xs` are close together and ordered from smallest to largest, then when we plot `xs` against `expanded_xs[:,j]` it gives the illusion of plotting a smooth curve representing $y=x^j$. If you set the value of `num_points` to a much smaller number, e.g. $5$, you will see that the illusion is no longer there and instead you see a number of connected straight lines.

$$dx \,dy =
