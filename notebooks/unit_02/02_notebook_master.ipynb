```python
# exercise 2.3 b)
from fomlads.model.regression import expand_to_monomials
from fomlads.model.regression import regularised_least_squares_weights
from fomlads.model.regression import construct_polynomial

from fomlads.evaluate.eval_regression import train_and_test_partition
from fomlads.evaluate.eval_regression import root_mean_squared_error

from fomlads.plot.predictions import plot_function_data_and_approximation


# create a filter for train and validation
train_filter = np.array([True if (i%2 == 0) else False for i in range(inputs.size)])

## TODO: implement random filter

# the validation filter is the converse of the train filter
valid_filter = ~train_filter

# apply the filters
inputs_train, targets_train, inputs_valid, targets_valid = train_and_test_partition(
    inputs, targets, train_filter, valid_filter)
print(f"inputs_train.shape = {inputs_train.shape}")
print(f"targets_train.shape = {targets_train.shape}")

# define degree of polynomial to fit
degree = 13
# convert training inputs into matrix of monomials
input_mtx_train = expand_to_monomials(inputs_train, degree)

# define regularisation parameters to evaluate
lambdas = np.logspace(-17,-8,11)
# create a 1d array for results one per lambda_
rmses = np.empty(lambdas.size)
# for each lambda_, fit regularised weights on training data
# and evaluate on validation data
for l, lambda_ in enumerate(lambdas):
    # find the weights that fit the data in a regularised squares way
    weights = regularised_least_squares_weights(input_mtx_train, targets_train, lambda_=lambda_)
    # use weights to create a function that takes inputs and returns predictions
    # in python, functions can be treated like variables
    approx_func = construct_polynomial(degree, weights)
    # apply the prediction function to the validation inputs
    preds_valid = approx_func(inputs_valid)
    # get the root mean squared error
    rmses[l] = root_mean_squared_error(targets_valid, preds_valid)

# plot the results
plt.figure()
plt.plot(lambdas, rmses)
plt.xscale('log')
plt.xlabel('$\lambda$')
plt.ylabel('RMSE')

#plt.ylim([None, 0.3])
```

Note that 1 standard error is equal to one (sample) standard deviation divided by the number of trials, so if your RMSE on fold $k$ is $x_k$, then your sample mean error is 
$$\bar{x} = \frac{1}{K}\sum_{k=1}^{K}x_k$$
your (sample) standard deviation is
$$\bar{s} = \sqrt{\frac{1}{K-1}\sum_{k=1}^{K}(x_k- \bar{x})^2}$$
and sample standard error is
$$\bar{e} = \frac{\bar{s}}{\sqrt{K}}$$
You can plot error bars with `matplotlib.pyplot.errorbar` where argument `yerr` takes the values of the standard error. Alternatively, you can use `matplotlib.pyplot.fill_between`, with an `alpha` value less than 1, to shade the region corresponding to $\bar{x} \pm \bar{e}$*
