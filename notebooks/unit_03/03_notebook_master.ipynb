**An optional question to attempt in your own time.**

Attempt Exercise 1.41 from (Bishop, 2006), repeated here for convenience:

> Using the sum and product rules of probability, show that the mutual
information I(x, y) satisﬁes the relation (1.121).

where Equation(1.121) is written as

$$
\mutualinfo{\vec{x}}{\vec{y}} 
= \entropy{\vec{x}} - \conditionalentropy{\vec{x}}{\vec{y}} 
= \entropy{\vec{y}} - \conditionalentropy{\vec{y}}{\vec{x}}
$$

Note that here $\vec{x}$ and $\vec{y}$ are taken to be random variables rather than outcomes of random variables. 

**A short note on notation**

Mutual information is expressed in Equation (1.120) as:

$$
\begin{aligned}
\mutualinfo{\vec{x}}{ \vec{y}}
& \equiv \kldiv{p(\vec{x}, \vec{y}) }{ p(\vec{x})p(\vec{y})}
\\
& = \int \int p(\vec{x}, \vec{y}) \ln \frac{p(\vec{x})p(\vec{y})}{p(\vec{x}, \vec{y})}
\, d\vec{x} \, d\vec{y}
\end{aligned}
$$

The problems with this notation become apparent with Equation (1.120) as here, for instance, $\vec{x}$ refers to a  random variable on the left hand side of the top line, but elsewhere it is the values that the RV can take. A more explicit notation would be to use $\vec{X}$ for the RV and $\vec{x}$ for outcomes to give:

$$
\begin{aligned}
\mutualinfo{\vec{X}}{ \vec{Y}}
& \equiv \kldiv{p_{\vec{X}\vec{Y}} }{ p_\vec{X} \cdot p_\vec{Y})}
\\
& = \int \int p_{\vec{X}\vec{Y}}(\vec{x}, \vec{y}) 
\ln \frac{p_\vec{X}(\vec{x})p_\vec{Y}(\vec{y})}{p_{\vec{X}\vec{Y}}(\vec{x}, \vec{y})}
\, d\vec{x} \, d\vec{y}
\end{aligned}
$$

A discussion on the notational issues with random variables and expectations etc can be found [here](https://statmodeling.stat.columbia.edu/2020/02/05/abuse-of-expectation-notation/). Hereafter, we will tend to use Bishop's (and others') notation, but it is helpful to understand why it can be ambiguous.

### 3.5 b)


The second term in Beta involving $\mu$ is sometimes called the **form** of the function as it determines the shape. Can you plot below the form of $\text{Beta}(\mu|a=2,b=2)$ (ignoring the normalisation)?

```python
# We want to plot the gamma function and overlay the factorial for
# values in the range (0,5]
# First let's get the values x at which to evaluate the
# gamma function (we drop the 0)
cs = np.linspace(0,5,102)
# let's import the appropriate library
# and  get the gamma values
import scipy.special
gs = scipy.special.gamma(cs+1)

# we also want just the integers in the range (0,5]
ns = np.arange(6)
# I will use the math.factorial function to evaluate the factorial
import math
facts = np.array([math.factorial(n) for n in ns])

#  let's plot the gamma values
plt.plot(cs, gs, label=r'$\Gamma(x+1)$')
# and overlay the factorials
plt.plot(ns, facts, 'o', label=r'$x!$')
plt.legend()
plt.xlabel('x')

```
